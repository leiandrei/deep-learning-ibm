```python
# demo rnn
from tensorflow import keras
from keras.utils import pad_sequences
from keras.models import Sequential
from keras.layers import Dense, Embedding
from keras.layers import SimpleRNN
from keras.datasets import imdb
from keras import initializers
```

```python
max_feats = 20000
max_length = 30
bacth_size = 32

(xtrain, ytrain), (xtest, ytest) = imdb.load_data(num_words=max_feats)
print(len(xtrain), 'train sequences')
print(len(xtest), 'test sequences')
```

```python
xtrain = pad_sequences(xtrain, maxlen=max_length)
xtest = pad_sequences(xtest, maxlen=max_length)

xtrain[123, :]
```

## Building RNN

```python
rnn_hidden_dim = 5
word_embedding_dim = 50

model_rnn = Sequential()

model_rnn.add(
    Embedding(max_feats, word_embedding_dim))
model_rnn.add(
    SimpleRNN(rnn_hidden_dim, kernel_initializer=initializers.RandomNormal(stddev=0.001),
              recurrent_initializer=initializers.Identity(gain=1.0),
              activation='relu',
              input_shape=xtrain.shape[1:]))
model_rnn.add(Dense(1, activation='sigmoid'))
model_rnn.summary()
```

```python
rmsprop = keras.optimizers.RMSprop(learning_rate= .0001)
model_rnn.compile(
    loss='binary_crossentropy',
    optimizer=rmsprop,
    metrics=['accuracy']
)

model_rnn.fit(
    xtrain, ytrain, batch_size=bacth_size, epochs=10, validation_data=(xtest, ytest)
)

score, acc = model_rnn.evaluate(xtest, ytest, bacth_size)
print(f"Test Score: {score}")
print(f"Test Accuracy: {acc}")
```
